```python
# 文件名: traditional_models_final.py

import pandas as pd
import numpy as np
import os

from sklearn.model_selection import StratifiedKFold
from sklearn.preprocessing import StandardScaler, LabelEncoder
from sklearn.metrics import classification_report

from sklearn.neighbors import KNeighborsClassifier
from sklearn.tree import DecisionTreeClassifier
from sklearn.svm import SVC
from sklearn.discriminant_analysis import LinearDiscriminantAnalysis
from sklearn.ensemble import RandomForestClassifier, GradientBoostingClassifier

from scipy.stats import skew, kurtosis
from sklearn.impute import SimpleImputer


# ===================================================================
# 步骤 1：全局配置
# ===================================================================

DATA_FILE_NAME = 'electrochem_data.csv'

SENSOR_ORDER = [
    'FGG',
    'FGB',
    'CP',
    'FETUB',
    'HBD',
    'CPN2',
    'AGT',
    'LUM',
    'ACTB'
]

CLASS_NAMES = ['W', 'Y', 'Z']

N_SPLITS = 5
NUM_FEATURES_TO_SELECT = 6
RANDOM_STATE = 64

# ===================================================================
# 步骤 2：特征提取函数
# ===================================================================

def extract_full_features(fingerprints, num_sensors):

    num_samples = fingerprints.shape[0]

    len_per_sensor = (
        fingerprints.shape[1] // num_sensors
    )

    feature_table = np.zeros(
        (
            num_samples,
            num_sensors * 2 * 6
        )
    )

    for i in range(num_samples):

        for j in range(num_sensors):

            segment = fingerprints[
                i,
                j * len_per_sensor:
                (j + 1) * len_per_sensor,
                :
            ]

            current_ch = segment[:, 0]
            voltage_ch = segment[:, 1]

            stats_current = [
                np.mean(current_ch),
                np.std(current_ch),
                np.max(current_ch),
                np.min(current_ch),
                skew(current_ch),
                kurtosis(current_ch)
            ]

            stats_voltage = [
                np.mean(voltage_ch),
                np.std(voltage_ch),
                np.max(voltage_ch),
                np.min(voltage_ch),
                skew(voltage_ch),
                kurtosis(voltage_ch)
            ]

            start_idx = j * 12

            feature_table[
                i,
                start_idx:start_idx + 12
            ] = (
                stats_current
                +
                stats_voltage
            )

    return feature_table


# ===================================================================
# 步骤 3：读取数据
# ===================================================================

print("\n正在加载数据...")


if not os.path.exists(DATA_FILE_NAME):

    raise FileNotFoundError(
        f"错误：未找到数据文件 '{DATA_FILE_NAME}'"
    )


df = pd.read_csv(
    DATA_FILE_NAME,
    encoding='utf-8'
)


df.columns = (
    df.columns
    .str.strip()
    .str.lower()
)


measurement_ids = (
    df['type']
    .unique()
)


X_combined = []
y_labels = []


current_cols = [
    col
    for col in df.columns
    if 'current_dot' in col
]


voltage_cols = [
    col.replace(
        'current',
        'voltage'
    )
    for col in current_cols
]


# ===================================================================
# 按测量编号整理9个传感器数据
# ===================================================================

for mid in measurement_ids:

    one_measurement_df = df[
        df['type'] == mid
    ]

    if (
        one_measurement_df.shape[0]
        != len(SENSOR_ORDER)
    ):

        continue


    sensor_signals = []
    valid_sample = True


    for sensor in SENSOR_ORDER:

        sensor_row = (
            one_measurement_df[
                one_measurement_df[
                    'sensor_type'
                ] == sensor
            ]
        )


        if sensor_row.empty:

            valid_sample = False
            break


        current = (
            sensor_row[
                current_cols
            ]
            .values
            .flatten()
        )


        voltage = (
            sensor_row[
                voltage_cols
            ]
            .values
            .flatten()
        )


        sensor_signals.append(
            np.stack(
                [
                    current,
                    voltage
                ],
                axis=1
            )
        )


    if not valid_sample:

        continue


    X_combined.append(
        np.concatenate(
            sensor_signals,
            axis=0
        )
    )


    y_labels.append(
        str(mid)[0].upper()
    )


# ===================================================================
# 转换为 NumPy
# ===================================================================

X_fingerprints_raw = np.array(
    X_combined
)

y_raw = np.array(
    y_labels
)


# ===================================================================
# 标签编码
# ===================================================================

label_encoder = (
    LabelEncoder()
    .fit(y_raw)
)


y_encoded = (
    label_encoder
    .transform(y_raw)
)


# ===================================================================
# 特征提取
# ===================================================================

X_features_full = (
    extract_full_features(
        X_fingerprints_raw,
        len(SENSOR_ORDER)
    )
)





# ===================================================================
# 固定随机抽取6个统计特征
# ===================================================================

np.random.seed(RANDOM_STATE)


selected_feature_indices = (
    np.random.choice(
        X_features_full.shape[1],
        NUM_FEATURES_TO_SELECT,
        replace=False
    )
)


X_features_selected = (
    X_features_full[
        :,
        selected_feature_indices
    ]
)




# ===================================================================
# 缺失值填补
# ===================================================================

imputer = SimpleImputer(
    strategy='mean'
)


X_features_imputed = (
    imputer.fit_transform(
        X_features_selected
    )
)


# ===================================================================
# 标准化
# ===================================================================

scaler_ml = StandardScaler()


X_features_scaled_ml = (
    scaler_ml.fit_transform(
        X_features_imputed
    )
)


print(
    "数据准备完成。"
)


# ===================================================================
# 步骤 4：建立传统机器学习模型
# ===================================================================

models_to_compare = {

    'LDA':

        LinearDiscriminantAnalysis(),


    'KNN':

        KNeighborsClassifier(
            n_neighbors=9
        ),


    'Decision Tree':

        DecisionTreeClassifier(
            max_depth=2,
            min_samples_leaf=4,
            random_state=RANDOM_STATE
        ),


    'Random Forest':

        RandomForestClassifier(
            n_estimators=30,
            max_depth=2,
            min_samples_leaf=3,
            random_state=RANDOM_STATE
        ),


    'SVM':

        SVC(
            kernel='rbf',
            C=0.2,
            gamma='scale',
            random_state=RANDOM_STATE
        ),


    'Gradient Boosting':

        GradientBoostingClassifier(
            n_estimators=30,
            learning_rate=0.03,
            max_depth=1,
            random_state=RANDOM_STATE
        )
}


# ===================================================================
# 步骤 5：5折分层交叉验证
# ===================================================================

all_model_results = {}


print(
    "\n正在进行 5-Fold Cross-Validation..."
)


for name, model in (
    models_to_compare.items()
):


    skf = StratifiedKFold(
        n_splits=N_SPLITS,
        shuffle=True,
        random_state=RANDOM_STATE
    )


    y_true_all = []
    y_pred_all = []


    for train_idx, val_idx in skf.split(
        X_features_scaled_ml,
        y_encoded
    ):


        X_train = (
            X_features_scaled_ml[
                train_idx
            ]
        )


        X_val = (
            X_features_scaled_ml[
                val_idx
            ]
        )


        y_train = (
            y_encoded[
                train_idx
            ]
        )


        y_val = (
            y_encoded[
                val_idx
            ]
        )


        # 训练模型
        model.fit(
            X_train,
            y_train
        )


        # 模型预测
        y_pred = model.predict(
            X_val
        )


        y_true_all.extend(
            y_val
        )


        y_pred_all.extend(
            y_pred
        )


    # ===================================================================
    # 性能指标
    # ===================================================================

    report = classification_report(
        y_true_all,
        y_pred_all,
        target_names=CLASS_NAMES,
        output_dict=True,
        zero_division=0
    )


    all_model_results[name] = {

        'Accuracy':
            report[
                'accuracy'
            ],

        'Precision':
            report[
                'weighted avg'
            ][
                'precision'
            ],

        'Recall':
            report[
                'weighted avg'
            ][
                'recall'
            ],

        'F1-Score':
            report[
                'weighted avg'
            ][
                'f1-score'
            ]
    }


# ===================================================================
# 步骤 6：打印最终结果
# ===================================================================

print(
    "\n"
    + "=" * 78
)


print(
    "Traditional Machine Learning Results"
)


print(
    "=" * 78
)


display_order = [
    'LDA',
    'SVM',
    'Random Forest',
    'Decision Tree',
    'KNN',
    'Gradient Boosting'
]


display_names = {

    'LDA':
        'LDA',

    'SVM':
        'SVM',

    'Random Forest':
        'RF',

    'Decision Tree':
        'DT',

    'KNN':
        'KNN',

    'Gradient Boosting':
        'GB'
}


print(
    f"{'Algorithm':<12}"
    f"{'Accuracy':<14}"
    f"{'Precision':<14}"
    f"{'Recall':<14}"
    f"{'F1-Score':<14}"
)


print(
    "-" * 68
)


for name in display_order:

    result = all_model_results[
        name
    ]

    print(

        f"{display_names[name]:<12}"

        f"{result['Accuracy'] * 100:<13.2f}%"

        f"{result['Precision']:<14.4f}"

        f"{result['Recall']:<14.4f}"

        f"{result['F1-Score']:<14.4f}"
    )


print(
    "=" * 78
)



print(
    "\n分析完成。"
)
```
